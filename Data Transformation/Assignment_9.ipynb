```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
df = pd.read_csv('adult_with_headers (1).csv')
```

```python
print(df.isnull().sum())
```

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

# Median for numerical (robust to outliers)
num_imputer = SimpleImputer(strategy='median')
df[num_cols] = num_imputer.fit_transform(df[num_cols])

# Mode for categorical (most frequent category)
cat_imputer = SimpleImputer(strategy='most_frequent')
df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])
```

```python
# Standard Scaling
scaler_standard = StandardScaler()
df_standard = df.copy()
df_standard[num_cols] = scaler_standard.fit_transform(df_standard[num_cols])
```

```python
# Min-Max Scaling
scaler_minmax = MinMaxScaler()
df_minmax = df.copy()
df_minmax[num_cols] = scaler_minmax.fit_transform(df_minmax[num_cols])
```

```python
print("Standard Scaled:\n", df_standard[num_cols].head())
print("\nMin-Max Scaled:\n", df_minmax[num_cols].head())
```

##**Scenarios for Each Scaling Technique**
###1. Standard Scaling:
- What it does: Transforms features to have mean
= 0 and standard deviation = 1.

  **Preferred when:**

   - Algorithms assume data is normally distributed (Linear Regression, Logistic Regression).
   - PCA (Principal Component Analysis), since it relies on variance.
   - Regularization methods (Ridge, Lasso) where scale affects penalty terms.
   - Why: Centers data and makes variance uniform, which stabilizes optimization.

###2. Min-Max Scaling:
- What it does: Rescales features to a fixed range (default [0,1]).
  
  **Preferred when:**
  - Distance-based algorithms (KNN, K-Means, SVM with RBF kernel).
  - Neural networks, where bounded inputs help faster convergence.
  - When you want to preserve relative relationships between values.
  - Why: Ensures all features contribute equally without one dominating due to scale.

```python
# One-Hot Encoding for categorical variables with < 5 categories
onehot_cols = [col for col in cat_cols if df[col].nunique() < 5]
df_onehot = pd.get_dummies(df, columns=onehot_cols, drop_first=True)

# Label Encoding for categorical variables with >= 5 categories
label_cols = [col for col in cat_cols if df[col].nunique() >= 5]
label_encoder = LabelEncoder()

for col in label_cols:
    df_onehot[col] = label_encoder.fit_transform(df_onehot[col])

print("\nData after encoding:\n", df_onehot.head())
```

###**One-Hot Encoding**
- Pros:
  - No ordinal relationship imposed (categories treated independently).
  - Works well for algorithms sensitive to categorical meaning (Linear Regression, Logistic Regression).
- Cons:
  - Increases dimensionality (curse of dimensionality).
  - Not efficient for high-cardinality features.
###**Label Encoding**
- Pros:
  - Compact representation (single column).
  - Useful for tree-based models (Decision Trees, Random Forests, XGBoost) that can handle categorical splits.
- Cons:
  - Imposes an artificial ordinal relationship which can mislead algorithms like Linear Regression or KNN.
  - Not suitable for non-tree models unless categories are truly ordinal.




#**3. Feature Engineering:**

```python
print(df.columns)
```

- Mapping these categories to years of education converts them into a meaningful numeric scale.
- Education level is strongly correlated with income, so this feature helps the model capture that relationship better.


```python
if 'education-num' in df.columns:
    df['Education_Num'] = df['education-num']

else:
    df['education'] = df['education'].str.strip()
    education_map = {'Preschool': 0, '1st-4th': 4, '5th-6th': 6, '7th-8th': 8,
        '9th': 9, '10th': 10, '11th': 11, '12th': 12,
        'HS-grad': 12, 'Some-college': 13, 'Assoc-voc': 14, 'Assoc-acdm': 14,
        'Bachelors': 16, 'Masters': 18, 'Doctorate': 20, 'Prof-school': 19
    }
    df['Education_Num'] = df['education'].map(education_map)
```

- People working more hours often earn more, but the effect is non-linear, so categorical grouping is useful.


```python
def work_hours_category(hours):
    if hours < 30:
        return "Part-time"
    elif hours <= 40:
        return "Full-time"
    else:
        return "Overtime"

df['Work_Hours_Category'] = df['hours_per_week'].apply(work_hours_category)
```

### Check skewness of numerical features

###Applying log transformation to skewed feature


```python
df['capital_gain_Log'] = np.log1p(df['capital_gain'])
```

###Comparing skewness before and after


```python
print("Skewness before:", df['capital_gain'].skew())
print("Skewness after log transform:", df['Capital_Gain_Log'].skew())
```

* **Education_Num:** Converts categorical education levels into a numeric scale (years of education), improving correlation with income.
- **Work_Hours_Category:** Groups continuous work hours into meaningful categories (Part-time, Full-time, Overtime).
- **Log Transformation on Capital-Gain:** Reduces skewness, stabilizes variance, and makes the distribution more model-friendly.
