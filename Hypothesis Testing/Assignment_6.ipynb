####1. Build 99% Confidence Interval Using Sample Standard Deviation

* Assuming the sample is representative of the population, construct a 99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation. Explain the steps you take and the rationale behind using the t-distribution for this task

```python
import numpy as np
import scipy.stats as stats
```

```python
# Inputing the sample data
durability = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
```

```python
# Calculating sample statistics
n = len(durability)
mean = np.mean(durability)
std_dev = np.std(durability, ddof=1)  # sample standard deviation
```

```python
#Determining the t-critical value for 99% confidence
confidence = 0.99
alpha = 1 - confidence
df = n - 1
t_critical = stats.t.ppf(1 - alpha/2, df)
```

```python
#Computing the margin of error
margin_error = t_critical * (std_dev / np.sqrt(n))
```

```python
# Constructing the confidence interval
lower_bound = mean - margin_error
upper_bound = mean + margin_error
```

```python
# Displaying results
print(f"Sample Mean: {mean:.4f}")
print(f"Sample Std Dev: {std_dev:.4f}")
print(f"t-Critical (99% CI, df={df}): {t_critical:.4f}")
print(f"Margin of Error: {margin_error:.4f}")
print(f"99% Confidence Interval: ({lower_bound:.4f}, {upper_bound:.4f})")
```

##**Interpretation**
* This confidence interval tells us that we are 99% confident the true mean durability of the print-heads lies within the calculated range. It reflects the uncertainty due to both the small sample size and the variability in the data.

#**2) HYPOTHESIS TESTING**

###Hypotheses Statement:

* Null Hypothesis (H0): The weekly operating cost follows the theoretical model (μ = ₹4,000)
* Alternative Hypothesis (H1): The weekly operating cost is higher than the theoretical model (μ > ₹4,000)

###Calculating the Test Statistic:

```python
#Use the following formula to calculate the test statistic(z):
#where:
#	ˉxˉ = sample mean weekly cost (Rs. 3,050)
#	μ = theoretical mean weekly cost according to the cost model (W = $1,000 + $5X for X = 600 units)
#	σ = 5*25 units
#	n = sample size (25 restaurants)
```

```python
import numpy as np
from scipy.stats import norm
```

```python
# Given values
sample_mean = 3050
X_mean = 600
X_std = 25
n = 25

# Theoretical mean weekly cost
mu = 1000 + 5 * X_mean
sigma = 5 * X_std

# Standard error
SE = sigma / np.sqrt(n)

# Test statistic (z-score)
z = (sample_mean - mu) / SE
print(z)
```

###Determine the Probability and Compare:

```python
#Using the alpha level of 5% (α = 0.05),
```

```python
# Alpha level
alpha = 0.05

# Critical value for one-tailed test at 5% significance
z_critical = norm.ppf(1 - alpha)

# Compare test statistic with critical value
print(f"Test Statistic (z): {z:.2f}")
print(f"Critical Value (z_critical): {z_critical:.2f}")
```

###Make a Decision:

* Compare the test statistic with the critical value to decide whether to reject the null hypothesis.

```python
if z > z_critical:
    decision = "Reject the null hypothesis"
else:
    decision = "Fail to reject the null hypothesis"

print(decision)
```

###Conclusion:

* Based on the decision in step 4, conclude whether there is strong evidence to support the restaurant owners' claim that the weekly operating costs are higher than the model suggests.

```python
if decision == "Reject the null hypothesis":
    conclusion = "There is strong evidence that the weekly operating costs are higher than the model suggests."
else:
    conclusion = "There is insufficient evidence to support the claim that weekly operating costs are higher."

print(conclusion)
```
